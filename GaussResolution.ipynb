```c++
import ROOT
%load_ext JupyROOT
```

```c++
%%cpp
const int NPART = 100000;
const float mean = 8.;
const float sigma = 3.;
```

```c++
%%cpp
  gDirectory->Clear();
  TList* canvases = (TList*)gROOT->GetListOfCanvases();
  TIter next_canvas(canvases);
  TObject* obj = nullptr;

  while((obj = next_canvas()))
  {
    TCanvas* c = dynamic_cast<TCanvas*>(obj);
    if(c)
    {
        delete c;
    }
  }
  gStyle->SetOptStat(0);
  TRandom3 *rand = new TRandom3(0);

  TH1F *Gauss = new TH1F("Gauss", ";Energy (GeV);Entries", 20, 0., 20.);
  Gauss->Sumw2();

  TF1 *Gauss_func = new TF1("Gauss_func", "gaus(0)", mean-(5.*sigma), mean+(5.*sigma));
  Gauss_func->SetParameters(1000./(sigma*sqrt(2.*M_PI)), mean, sigma);

  for(int i = 0; i < NPART; i++)
  {
    float energy = rand->Gaus(mean, sigma);

    Gauss->Fill(energy);
  }

  TCanvas *cGauss = new TCanvas("cGauss", "cGauss", 550, 550);
  cGauss->cd();
  gPad->SetLeftMargin(0.15);
  gPad->SetRightMargin(0.05);
  Gauss->SetMarkerStyle(kFullCircle);
  Gauss->SetMarkerColor(kBlack);
  Gauss->SetLineWidth(2);
  Gauss->SetLineColor(kBlack);
  Gauss->Draw();
  cGauss->Draw();

  TCanvas *cGaussFit = new TCanvas("cGaussFit", "cGaussFit", 550, 550);
  cGaussFit->cd();
  gPad->SetLeftMargin(0.15);
  gPad->SetRightMargin(0.05);
  Gauss->SetMarkerStyle(kFullCircle);
  Gauss->SetMarkerColor(kBlack);
  Gauss->SetLineWidth(2);
  Gauss->SetLineColor(kBlack);
  Gauss->Fit(Gauss_func, "WL", "0", 0., 20.);
  Gauss->GetFunction("Gauss_func")->SetBit(TF1::kNotDraw);
  Gauss->Draw();
  Gauss_func->Draw("same");
  cGaussFit->Draw();
```
