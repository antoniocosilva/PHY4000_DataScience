```c++
int NDICE = 1;
int NROLLS = 10;

bool print_histo_entries = true;
```

```c++
float GetMean(TH1F *histo)
{
  float mean = 0.;

  for(int i = 1; i <= histo->GetXaxis()->GetNbins(); i++)
  {
    mean += histo->GetBinContent(i)*histo->GetBinCenter(i);
  }

  return mean/histo->GetEntries();
}
```

```c++
float GetStdDev(TH1F *histo)
{
  float sDev = 0.;
  float mean = GetMean(histo);

  for(int i = 1; i <= histo->GetXaxis()->GetNbins(); i++)
  {
    sDev += histo->GetBinContent(i)*pow(histo->GetBinCenter(i) - mean, 2.);
  }

  return sqrt(sDev/(histo->GetEntries()-1));
}
```

```c++
gStyle->SetOptStat(0);
  TRandom3 *rand = new TRandom3(0);

  const int nbins = 6*NDICE - (NDICE - 1);
  const double lower_bin = 0.5 + (NDICE - 1.);
  const double upper_bin = 0.5 + 6.*NDICE;

  TProfile *DiceStat = new TProfile("DiceStat", "", 1, -0.5, 0.5);


  TH1F *hRolls = new TH1F("hRolls", ";Die Roll Value;Frequency", nbins, lower_bin, upper_bin);

  for(int i = 0; i < NROLLS; i++)
  {
    int comb_value = 0.;
    for(int n = 0; n < NDICE; n++)
    {
      int value = std::ceil(rand->Uniform(0.,6.));
      comb_value += value;
    }

    hRolls->Fill(comb_value);
    DiceStat->Fill(0., comb_value);
  }

  float mean = GetMean(hRolls);
  float sDev = GetStdDev(hRolls);

  std::cout << "Mean: " << mean << std::endl;
  std::cout << "Std Dev: " << sDev << std::endl;

  if(print_histo_entries)
  {
    for(int i = 1; i <= hRolls->GetXaxis()->GetNbins(); i++)
    {
      std::cout << "Bin center: " << hRolls->GetBinCenter(i) << " Entries: " << hRolls->GetBinContent(i) << std::endl;
    }
  }





  TCanvas *cRolls = new TCanvas("cRolls", "cRolls", 550, 550);
  cRolls->SetLeftMargin(0.15);
  hRolls->SetMinimum(0.);
  hRolls->SetLineWidth(2);
  hRolls->SetLineColor(kBlack);
  hRolls->Draw();
  cRolls->Draw();
```
