```c++
import ROOT
%load_ext JupyROOT
```

```c++
%%cpp
const int NPART = 300;
const float DET_EFF = 0.8;
```

```c++
%%cpp
  gDirectory->Clear();
  TList* canvases = (TList*)gROOT->GetListOfCanvases();
  TIter next_canvas(canvases);
  TObject* obj = nullptr;

  while((obj = next_canvas()))
  {
    TCanvas* c = dynamic_cast<TCanvas*>(obj);
    if(c)
    {
        delete c;
    }
  }
  
  gStyle->SetOptStat(0);
  TRandom3 *rand = new TRandom3(0);

  TH1F *GenEnergy = new TH1F("GenEnergy", "Particle Energy;Energy (GeV);Entries", 10, 0., 10.);
  GenEnergy->Sumw2();
  TH1F *RecEnergy = new TH1F("RecEnergy", "Particle Energy;Energy (GeV);Entries", 10, 0., 10.);
  RecEnergy->Sumw2();

  for(int i = 0; i < NPART; i++)
  {
    float prob = rand->Uniform(0., 1.);
    float energy = rand->Uniform(0., 10.);
    if(prob < DET_EFF)
    {
      RecEnergy->Fill(energy);
    }
    GenEnergy->Fill(energy);
  }

  TCanvas *cEnergy = new TCanvas("cEnergy", "cEnergy", 550, 550);
  cEnergy->cd();
  GenEnergy->SetMinimum(0.);
  GenEnergy->SetLineColor(kBlue);
  GenEnergy->SetLineWidth(2);
  GenEnergy->Draw("histo");
  RecEnergy->SetLineColor(kRed);
  RecEnergy->SetLineWidth(2);
  RecEnergy->Draw("same histo");
  cEnergy->Draw();

  TCanvas *cEnergy2 = new TCanvas("cEnergy2", "cEnergy2", 550, 550);
  cEnergy2->cd();
  GenEnergy->SetMinimum(0.);
  GenEnergy->SetLineColor(kBlue);
  GenEnergy->SetLineWidth(2);
  GenEnergy->Draw();
  RecEnergy->SetLineColor(kRed);
  RecEnergy->SetLineWidth(2);
  RecEnergy->Draw("same");
  cEnergy2->Draw();

  TCanvas *cEff = new TCanvas("cEff", "cEff", 1100, 550);
  cEff->Divide(2,1);
  cEff->cd(1);
  TH1F *Efficiency = (TH1F*)RecEnergy->Clone("Efficiency");
  Efficiency->SetTitle("Efficiency;Energy (GeV);Efficiency");
  Efficiency->GetYaxis()->SetRangeUser(0., 1.);
  Efficiency->SetLineColor(kRed);
  Efficiency->SetLineWidth(2);
  Efficiency->Divide(GenEnergy);
  Efficiency->Draw();
  
  cEff->cd(2);
  TH1F *Efficiency_Binomial = (TH1F*)RecEnergy->Clone("Efficiency_Binomial");
  Efficiency_Binomial->SetTitle("Efficiency (Binomial);Energy (GeV);Efficiency");
  Efficiency_Binomial->SetMinimum(0.);
  Efficiency_Binomial->SetLineColor(kRed);
  Efficiency_Binomial->SetLineWidth(2);
  Efficiency_Binomial->Divide(RecEnergy, GenEnergy, 1., 1., "B");
  Efficiency_Binomial->GetYaxis()->SetRangeUser(0., 1.);
  Efficiency_Binomial->Draw();
  cEff->Draw();
```
