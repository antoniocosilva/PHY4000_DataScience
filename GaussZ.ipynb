```c++
import ROOT
%load_ext JupyROOT
```

```c++
%%cpp
const int NPART = 10000;

const float sigma = 3.;
const float mean = 10.;
const float h0_mean = 10.;
```

```c++
%%cpp
    gStyle->SetOptStat(0);
    TRandom3 *rand = new TRandom3(0);
    
    TH1F *Gauss = new TH1F("Gauss", ";Energy (GeV);Entries", 20, 0., 20.);
    Gauss->Sumw2();

    TH1F *GaussZ = new TH1F("GaussZ", ";Z;Entries", 12, -6., 6.);
    GaussZ->Sumw2();
    
    TF1 *Gauss_func = new TF1("Gauss_func", "gaus(0)", mean-(5.*sigma), mean+(5.*sigma));
    Gauss_func->SetParameters(1000./(sigma*sqrt(2.*M_PI)), mean, sigma);

    TF1 *GaussZ_func = new TF1("GaussZ_func", "gaus(0)", -10, 10);
    GaussZ_func->SetParameters(1000./(sigma*sqrt(2.*M_PI)), 0., 1.);
    
    for(int i = 0; i < NPART; i++)
    {
        float energy = rand->Gaus(mean, sigma);
        Gauss->Fill(energy);
        GaussZ->Fill((energy-h0_mean)/sigma);
    }
    
    TCanvas *cGauss = new TCanvas("cGauss", "cGauss", 2200, 1100);
    cGauss->Divide(2,1);
    cGauss->cd(1);
    gPad->SetLeftMargin(0.15);
    gPad->SetRightMargin(0.05);
    Gauss->SetMarkerStyle(kFullCircle);
    Gauss->SetMarkerColor(kBlack);
    Gauss->SetLineWidth(2);
    Gauss->SetLineColor(kBlack);
    Gauss->Fit(Gauss_func, "", "0", 0., 20.);
    Gauss->GetFunction("Gauss_func")->SetBit(TF1::kNotDraw);
    Gauss->Draw();
    Gauss_func->Draw("same");
    
    cGauss->cd(2);
    gPad->SetLeftMargin(0.15);
    gPad->SetRightMargin(0.05);
    GaussZ->SetMarkerStyle(kFullCircle);
    GaussZ->SetMarkerColor(kBlack);
    GaussZ->SetLineWidth(3);
    GaussZ->SetLineColor(kBlack);
    //GaussZ->Scale(1./GaussZ->Integral());
    GaussZ->SetMaximum(GaussZ->GetMaximum()*1.2);
    GaussZ->Fit(GaussZ_func, "", "0", -10., 10.);
    GaussZ->GetFunction("GaussZ_func")->SetBit(TF1::kNotDraw);
    GaussZ_func->SetLineWidth(3);
    GaussZ->Draw();
    GaussZ_func->Draw("same");
    cGauss->Draw();
```
