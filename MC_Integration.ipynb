```python
%%cpp
const long long NENTRIES = 1000;
```

```python
%%cpp
    TRandom3 *rand = new TRandom3(0);

    //Welford's variables
    long long N = 0;
    double mean = 0.0;
    double M2 = 0.0;
    

    for(int i = 0; i < NENTRIES; i++)
    {
        double x = rand->Uniform(0.,1.);
        double f = TMath::Exp(-TMath::Power(x, 2));

        //Apply Welford's updates
        N++;
        double delta = f - mean;
        mean += delta / N;
        double delta2 = f - mean;
        M2 += delta * delta2;
    }

    double variance = (N > 1) ? M2 / (N - 1) : 0.0; //Use (N - 1) for sample variance
    double sigma = std::sqrt(variance/N);


    std::cout << "Used " << NENTRIES << " random numbers for Monte Carlo calculation" << std::endl;

    std::cout << "Integral by Monte Carlo: " << mean << "(" << sigma << ")" << std::endl;

    TF1 *my_exp = new TF1("my_exp", "exp(-x*x)", 0., 1.);

    std::cout << "Integral by ROOT: " << my_exp->Integral(0.,1.) << std::endl;
```
