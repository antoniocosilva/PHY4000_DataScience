```python
%%cpp
const float BR_Cs_to_mBa = 0.946;
const float BR_Ba_to_Photon = 0.851;

const float lambda_Cs = log(2)/9.522E8; //s^-1
const float lambda_Ba = log(2)/2.55; //s^-1

const float Activity_Cs = 0.037; //1 pCi = 0.037 Bq

const float TimeInterval = 1000.;
```

```python
%%cpp
    TRandom3 *rand = new TRandom3(0);

    int N0 = std::ceil(Activity_Cs/lambda_Cs);

    std::cout << "N0 t_Cs: " << N0 << " Cs nuclei" << std::endl;

    int Cs_Decay_to_mBa = 0;
    int Cs_Decay_to_Ba = 0;
    int mBa_Decay_to_Ba = 0;

    for(int i = 0; i < N0; i++)
    {
        float t_Cs = -log(rand->Uniform(0., 1.))/lambda_Cs;

        if(t_Cs < TimeInterval)
        {
            float Decay_to_mBa = rand->Uniform(0., 1.);

            if(Decay_to_mBa < BR_Cs_to_mBa)
            {
                Cs_Decay_to_mBa++;
                float t_Ba = -log(rand->Uniform(0., 1.))/lambda_Ba;
                if(t_Ba < (TimeInterval-t_Cs))
                {
                    mBa_Decay_to_Ba++;
                }
            }
            else
            {
                Cs_Decay_to_Ba++;
            }
            
        }

    }

    std::cout << "Number of Cs decays: " << (Cs_Decay_to_Ba+Cs_Decay_to_mBa) << std::endl;
    std::cout << "Number of Cs decays to mBa: " << Cs_Decay_to_mBa << std::endl;
    std::cout << "Number of Cs decays to Ba: " << Cs_Decay_to_Ba << std::endl;
    std::cout << "Number of mBa decays to Ba: " << mBa_Decay_to_Ba << std::endl;
```
