```c++
import ROOT
%load_ext JupyROOT
```

```c++
%%cpp
const int NTIMES = 3000;
```

```c++
%%cpp
float GetMean(TH1F *histo)
{
  float mean = 0.;

  for(int i = 1; i <= histo->GetXaxis()->GetNbins(); i++)
  {
    mean += histo->GetBinContent(i)*histo->GetBinCenter(i);
  }

  return mean/histo->GetEntries();
}
```

```c++
%%cpp
  gDirectory->Clear();
  TObject *obj = gROOT->FindObject("cTimes");
  if(obj)
  {
      delete obj;
  }
  gStyle->SetOptStat(0);
  TRandom3 *rand = new TRandom3(0);

  TH1F *nTimes = new TH1F("nTimes", "X = 1, 2, 3; Value; Probability", 3, -0.5, 2.5);

  for(int i = 0; i < NTIMES; i++)
  {
    float prob = rand->Uniform(0., 1.);

    if(prob < 0.2)
    {
      nTimes->Fill(0.);
    }
    else if(prob > 0.2 && prob < 0.7)
    {
      nTimes->Fill(1.);
    }
    else if(prob > 0.7)
    {
      nTimes->Fill(2.);
    }
  }

  TCanvas *cTimes = new TCanvas("cTimes", "cTimes", 550, 550);
  cTimes->cd();
  nTimes->Scale(1./NTIMES);
  nTimes->GetYaxis()->SetRangeUser(0., 1.);
  nTimes->SetLineColor(kBlack);
  nTimes->SetLineWidth(2);
  nTimes->Draw("histo");

  float mean = nTimes->GetMean();

  TLine *line = new TLine(mean, 0., mean, 0.7);
  line->SetLineColor(kRed);
  line->SetLineWidth(2);
  line->Draw("same");

  TLine *line_exp_value = new TLine(1.1, 0., 1.1, 0.7);
  line_exp_value->SetLineColor(kBlue);
  line_exp_value->SetLineWidth(2);
  line_exp_value->Draw("same");

  TLegend *leg = new TLegend(0.55, 0.75, 0.89, 0.89);
  leg->SetFillColor(kWhite);
  leg->SetBorderSize(0);
  leg->AddEntry(line_exp_value, "Expected value E(X)", "l");
  leg->AddEntry(line, "Sample mean #bar{x}", "l");
  leg->Draw("same");
  cTimes->Draw();
```
