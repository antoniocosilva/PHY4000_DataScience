```python
%%cpp
const long long NENTRIES = 100000000;

const double R = 1.;
const double r = 0.2;
```

```python
%%cpp
    gStyle->SetOptStat(0);
    TRandom3 *rand = new TRandom3(0);

    long long Nhit = 0;

    TH2F *XZhits = new TH2F("XZhits", ";x (cm);z (cm)", 100, -0.3, 0.3, 100, -0.3, 0.3);

    TH1F *Rhits = new TH1F("Rhits", "Flight distance;Distance (m);Entries", 100, 1., 1.025);

    for(int i = 0; i < NENTRIES; i++)
    {
        double phi = 2.*TMath::Pi()*rand->Uniform(0., 1.);
        double cos_theta = 2.*rand->Uniform(0., 1.) - 1.;
        double sin_theta = sqrt(1.-TMath::Power(cos_theta, 2.));

        double ux = cos(phi)*sin_theta; //ux is the x-direction
        double uy = sin(phi)*sin_theta; //uy is the y-direction
        double uz = cos_theta; //uz is the z-direction

        if(uy < 0.)
        {
            continue;
        }

        double rho = R/uy; //rho is the radius in the x-z plane

        double x = rho*ux;
        double z = rho*uz;

        if((x*x + z*z) < r*r)
        {
            Nhit++;
            XZhits->Fill(x,z);
            Rhits->Fill(rho);
        }
    }

    std::cout << "Detector geometric acceptance: " << 100.*double(Nhit)/NENTRIES << "%" << std::endl;

    TCanvas *cXZ = new TCanvas("cXZ", "cXZ", 1100, 550);
    cXZ->Divide(2,1);
    cXZ->cd(1);
    gPad->SetRightMargin(0.15);
    XZhits->Draw("colz");
    cXZ->cd(2);
    gPad->SetLeftMargin(0.15);
    Rhits->Draw();
    cXZ->Draw();
```
